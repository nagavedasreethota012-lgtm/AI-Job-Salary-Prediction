### Deployed Application Link

The AI Job Salary Prediction application was developed using Streamlit Community Cloud.

**Deployed Application Link:** https://ai-job-salary-prediction-9znnwqeho5dep9lwe2razc.streamlit.app/

The application allows users to enter job-related details, such as job title, experience level, employment type, company location, company size, and remote work ratio, to predict the estimated salary of AI professionals using a trained Machine Learning regression model.

The application preprocesses the user inputs, converts categorical features into numerical representations, and passes the processed data to the trained model to generate salary predictions.

The predicted salary is displayed through an interactive Streamlit interface, helping users understand estimated salary ranges based on different job-related factors.

The project demonstrates the practical implementation of Machine Learning, data preprocessing, model deployment, and interactive web application development using Python and Streamlit.
