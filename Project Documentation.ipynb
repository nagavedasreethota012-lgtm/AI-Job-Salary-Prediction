# AI Job Salary Prediction

## 1. Problem Statement

The objective of this project is to predict job salaries using Machine Learning regression algorithms.

Salary prediction helps understand how different job-related factors influence salary levels. The project uses historical job salary data to train machine learning models and predict salaries based on input features.

## 2. Approach

The project follows these steps:

**Step 1: Data Collection**
- Load the job salary dataset.
- Understand the dataset structure and features.

**Step 2: Data Preprocessing**
- Check missing values.
- Handle categorical and numerical features.
- Prepare the dataset for model training.

**Step 3: Exploratory Data Analysis (EDA)**
- Analyze the dataset.
- Understand relationships between features and salary.
- Identify useful patterns.

**Step 4: Model Building**

Train and evaluate the following Machine Learning regression models:

- Linear Regression
- Decision Tree Regressor
- Random Forest Regressor
- Gradient Boosting Regressor

**Step 5: Model Evaluation**

Evaluate model performance using:

- Mean Absolute Error (MAE)
- Mean Squared Error (MSE)
- Root Mean Squared Error (RMSE)
- R² Score

**Step 6: Experiment Tracking**

Use MLflow and DagsHub to:
- Track model experiments.
- Log evaluation metrics.
- Compare different models.
- Store the best model as an artifact.

**Step 7: Best Model Selection**

Compare all trained models and select the best-performing model based on evaluation metrics.
